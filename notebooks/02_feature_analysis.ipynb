# Feature Engineering Analysis

Assess engineered features for modeling traffic congestion.

```python
import pandas as pd
from pathlib import Path

PROCESSED_PATH = Path('data/processed')
parquet_files = list(PROCESSED_PATH.glob('**/*.parquet'))
parquet_files[:5]
```

```python
if parquet_files:
    df = pd.read_parquet(parquet_files[0])
    df.describe().T
else:
    print('No processed parquet files found. Run the PySpark ETL first.')
```

```python
if parquet_files:
    feature_summary = df[[
        "vehicle_count",
        "volume_per_minute",
        "duration_minutes",
        "congestion_index",
        "weekpart_congestion_ratio"
    ]].describe().T
    feature_summary
```

### Interpretation Notes

- `feature_summary` highlights central tendency and dispersion of engineered metrics.
- Heatmaps reveal intra-day congestion surges; cross-reference with anomaly outputs.
- Scatter plot indicates which boroughs experience weekend-heavy congestion ratios.
