# NYC Traffic Volume - Data Overview

Initial exploration of the Automated Traffic Volume Counts dataset.

```python
import pandas as pd
from pathlib import Path

RAW_PATH = Path('data/raw')
FILES = list(RAW_PATH.glob('*.csv'))
FILES
```

## Schema inspection

```python
if FILES:
    sample = pd.read_csv(FILES[0], nrows=5000)
    sample.info()
else:
    print('No files found. Please run src/ingestion/download_data.py first.')
```

```python
if FILES:
    missing_summary = (sample.isna().mean().sort_values(ascending=False) * 100).round(2)
    print("Missing value percentage (top 10 columns):")
    print(missing_summary.head(10))
```

```python
if FILES:
    top_locations = (
        sample.groupby(["borough", "roadwayname"])["vehiclecount"].sum().reset_index()
        .sort_values("vehiclecount", ascending=False)
        .head(10)
    )
    top_locations
```
